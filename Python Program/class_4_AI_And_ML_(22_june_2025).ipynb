# **Check Number of Dimensions?**
NumPy Arrays provides the ndim attribute that returns an integer that tells us how many dimensions the array have.

```python
import numpy as np

a = np.array(42)
b = np.array([1, 2, 3, 4, 5])
c = np.array([[1, 2, 3], [4, 5, 6]])
d = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```

```python
a = np.array(42)
b =np.array([1,2,3,4,5])
c = np.array([[1,2,3], [4,6,7]])
d = np.array([[[1,5,2], [4,5,6], [4,8,9], [9,5,2]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```

# **Higher Dimensional Arrays**
An array can have any number of dimensions.

When the array is created, you can define the number of dimensions by using the ndmin argument.

Create an array with 5 dimensions and verify that it has 5 dimensions:

```python
import numpy as np

arr = np.array([1, 2, 3, 4], ndmin=5)

print(arr)
print('number of dimensions :', arr.ndim)
```

```python
def calculate_operations():
    # Get user input
    numbers_input = input("Enter two or more numbers separated by spaces: ")

    # Convert input string into a list of numbers
    numbers = list(map(float, numbers_input.split()))

    # Check if user entered at least two numbers
    if len(numbers) < 2:
        print("Please enter at least two numbers.")
        return

    # Perform operations
    total_sum = sum(numbers)

    # Subtraction: start from the first number and subtract the rest
    total_subtract = numbers[0]
    for num in numbers[1:]:
        total_subtract -= num

    # Multiplication
    total_multiply = 1
    for num in numbers:
        total_multiply *= num

    # Print results
    print("Sum:", total_sum)
    print("Subtraction:", total_subtract)
    print("Multiplication:", total_multiply)

# Example usage
calculate_operations()
```
