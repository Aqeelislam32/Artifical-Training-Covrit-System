# **A for loop is used for iterating over a sequence (that is either a list, a tuple, a dictionary, a set, or a string).**

This is less like the for keyword in other programming languages, and works more like an iterator method as found in other object-orientated programming languages.

With the for loop we can execute a set of statements, once for each item in a list, tuple, set etc.

```python
freind =['Ali','Asad','Saad' ]
for fiend in  freind:
  print('Happy New Year:',fiend)
print('Done')
```

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)
  if x == "banana":
    break
```

```python
x =[3,41,12,8,74,15]
for i in x:
  if i > 73:
    print('Large number',i)
print('Done')
```

# **Print Largest Value**

```python
x = [3,41,12,9,74,15]
larger = x[0]

for i in x:
  if i > larger:
    larger = i
print('larger number',larger)
```

The pop() method is used to remove an element from a list at a specified index and return that element. If no index is provided, it will remove and return the last element by default. This method is particularly useful when we need to manipulate a list dynamically, as it directly modifies the original list

```python
a = [10, 20, 30, 40]

# Remove the last element from list
a.pop()

print(a)
```
